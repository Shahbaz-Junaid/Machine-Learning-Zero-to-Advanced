# 🔹 Fine Tuning Kya Hoti Hai?

```python
# ✅ Roman Urdu:

# Fine Tuning ka matlab hai:

# Model ke parameters (settings) ko adjust karna taake best performance mile.
# Yani model ko train karne ke baad hum uski settings change karke usay aur better banate hain.

# ✅ English:
# Fine Tuning means adjusting the hyperparameters of a model to improve its performance.

# It helps in:
# Improving accuracy
# Reducing overfitting
# Getting optimal results
```

```python
# 🔥 Important Term: Hyperparameters
# Hyperparameters woh settings hoti hain jo model train hone se pehle set ki jati hain.

# Example (Logistic Regression):

# C
# max_iter
# penalty

# Example (Decision Tree):

# max_depth
# min_samples_split
```

```python
# manually way
```

```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score

# Load dataset
iris = load_iris()
X = iris.data
y = iris.target

# Different C values try karte hain
for c_value in [0.1, 1, 10, 100]:
    model = LogisticRegression(C=c_value, max_iter=200)
    scores = cross_val_score(model, X, y, cv=5)
    print("C =", c_value, "Average Accuracy =", scores.mean())
```

```python
# 🔥 Automatic Fine Tuning (GridSearchCV)
```

```python
from sklearn.model_selection import GridSearchCV

# Parameter grid define karo
param_grid = {
    'C': [0.1, 1, 10, 100]
}

# Model
model = LogisticRegression(max_iter=200)

# Grid Search
grid = GridSearchCV(model, param_grid, cv=5)

grid.fit(X, y)

print("Best Parameter:", grid.best_params_)
print("Best Score:", grid.best_score_)
```
